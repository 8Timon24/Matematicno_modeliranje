**Moving average model with $n$ delays**

We have an unknown system, which to an input signal $x(t)$ responds with output signal $y(t)$: $$x(t) \longrightarrow \boxed{\textrm{unknown system}}\longrightarrow y(t)\textrm{. }$$

Based on observations, say $N+1$ measurements, we would like to predict the behaviour of the system. Suppose that measurements of the signals 
$x(t)$ and $y(t)$ are taken at times $t = 0, 1, \ldots, N$. According to our model, the output $y(t)$ at time $t$ can be expressed as a linear 
combination of $n$ inputs $x(t), x(t-1), \ldots, x(t - n  + 1)$ from the time interval $[t-n+1,t]$, i.e.,
$$
	y(t) = h_1 x(t-n+1) + h_2 x(t-n+2) + \cdots + h_{n-1}x(t-1) + h_{n}x(t) \textrm{. }
$$
To be more precise: Our data consists of $N+1$ inputs
$$
	x_0, x_1, \ldots, x_N
$$
and $N+1$ corresponding outputs
$$
	y_0, y_1, \ldots, y_N \textrm{, }
$$
which are taken at integer times $t$ from the interval $[0, N]$. (With this notation we have $x_i = x(i)$ and $y_i = y(i)$.) 
Equation above becomes
$$
	y_k = h_1 x_{k-n+1} + h_2 x_{k-n+2} + \cdots + h_{n-1}x_{k-1} + h_{n}x_k \textrm{, }
$$
and we have such an equation for all $k = n-1, n, \ldots, N$. Thus, we have a system of $N-n+1$ linear equations with $n$ unknowns 
$h_1, \ldots, h_n$ where we assume that $N-n+1 \ge n$. (Actually, we assume that $N-n+1$ is much larger than $n$.) 
Let $A$ be the matrix of this system, $\mathbf{h} = [h_1, \ldots, h_n]^\mathsf{T}$ the vector of unknowns, and $\mathbf{y}$ the right-hand side. 
We are looking for the linear least squares solution of the system $A\mathbf{h} = \mathbf{y}$.

- Express the matrix $A$ of this linear system.
- Suppose $\mathbf{h}$ is known. How would you predict the output $y_0, \ldots, y_m$ of the system given input data $x_0, \ldots, x_m$? Which components of the output can be predicted?

Write a _Julia_ function `movavg(x, y, n)` which, given input data `x = [x0, ..., xN]` and output data `y = [y0, ..., yN]`, returns the coefficients `h = [h1, ..., hn]` for the moving average model with $n$ delays. (Specifications: `x`, `y`, and `h` are _row vectors_.)

```julia
# Generate some artificial test data (pick your own x, n, h and generate y). (Alternatively, use pre-generated data below.)
# Then solve the corresponding linear system and confirm that you get h you picked.
xt = [1, 2, 3, 4, 5, 6, 7, 8, 10]
yt = [10, 11, 12, 13, 14, 15, 16, 17, 18]
n = 5
```

```julia
# the function 'movavg'...
"""x so vhodni podatki v vrsticnem vektorju dolzine N+1, y so izhodni podatki v vrsticnem vektorju dolzine N+1. Funkcija movavg 
vrne vrsticni vektor h velikosti n, ki je vektor koeficientou za model"""
function movavg(x, y, n)
    N = length(x)-1;
    A = zeros(N-n+2, n) #inicializiramo matriko sistema
    
    for i in 1:(N-n+2)
        A[i, :] = x[i:i+n-1] #napolnemo matriko z x-i
    end
    
    h = A\y[n:N+1] #stolpicni
    return h'  #vrsticni
end

#movavg(xt, yt, n)
```

Suppose that all inputs and all outputs are constant, i.e., $x_i = y_i = 1$. First guess and then derive the solution $\mathbf{h}$ for arbitrary $N$ and $n$. Compare your solution with the output of `movavg`.

Write a _Julia_ function `prediction(x, h)` which, given input data `x` and coefficients `h`, returns the output `y` using our model.

```julia
# the function 'prediction'...
"""We assume x and y are constants, therefore x[i] = c for i = 1,...,N,  """
function prediction_const(x, h)
    hsum = sum(h);
    cx = x[1];
    y = [cx*hsum for i in 1:length(x)]
    return y
end

"""Prediction predicts the value of the function y(t) = h1*x1(t) + ... + hn*xn(t) given the coefficients h and input values x"""
function prediction(x, h)
    n = length(h)
    N = length(x)-1;
    y = zeros(N-n+2);
    y = [sum(h.* x[i:i+n-1]) for i in 1:N-n+2]
    return y
end
```

```julia
x1 = [1111, 2121, 322, 140, 1923, 230]
h1 = [2, 3, 4, 5, 6]
y1 = prediction(x1, h1)
println(y1)
movavg(x1, y1, 5)
```
